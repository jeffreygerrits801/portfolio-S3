```python
import numpy as np
import matplotlib.pyplot as plt
from torch.utils.tensorboard import SummaryWriter

units = [64, 128, 256, 512]
epochs = [3, 10]

val_loss = np.array([
    [0.39, 0.35],  # 64
    [0.39, 0.34],  # 128
    [0.39, 0.32],  # 256
    [0.37, 0.35],  # 512
])

train_loss = np.array([
    [0.37, 0.26],
    [0.34, 0.24],
    [0.32, 0.22],
    [0.32, 0.22],
])

val_improvement = val_loss[:,0] - val_loss[:,1]          
gen_gap_10 = val_loss[:,1] - train_loss[:,1]             

writer = SummaryWriter(log_dir="modellogs/summary_manual")


fig, ax = plt.subplots()
im = ax.imshow(val_loss, aspect="auto")
ax.set_xticks(range(len(epochs)))
ax.set_xticklabels(epochs)
ax.set_yticks(range(len(units)))
ax.set_yticklabels(units)
ax.set_xlabel("Epochs")
ax.set_ylabel("Units")
ax.set_title("Final validation loss (lower is better)")
for i in range(len(units)):
    for j in range(len(epochs)):
        ax.text(j, i, f"{val_loss[i,j]:.2f}", ha="center", va="center")
fig.colorbar(im, ax=ax)
writer.add_figure("summary/val_loss_heatmap", fig, global_step=0)
plt.close(fig)

fig, ax = plt.subplots()
ax.bar([str(u) for u in units], val_improvement)
ax.set_xlabel("Units")
ax.set_ylabel("Δ validation loss (3 epoch to 10 epoch)")
ax.set_title("Validation improvement when increasing epochs (3 → 10)")
writer.add_figure("summary/val_loss_improvement", fig, global_step=0)
plt.close(fig)


fig, ax = plt.subplots()
ax.bar([str(u) for u in units], gen_gap_10)
ax.set_xlabel("Units")
ax.set_ylabel("Validation loss - training loss (10 epochs)")
ax.set_title("Generalization gap at 10 epochs (higher may indicate overfitting risk)")
writer.add_figure("summary/generalization_gap_10", fig, global_step=0)
plt.close(fig)

writer.close()
print("Logged summary visuals to: modellogs/summary_manual")
```
